### Tabel metrik per dataset

```python
import pandas as pd
df = pd.read_csv(EXP / "eval_results.csv")
df["dataset"] = df["dataset"].astype(str)
for ds in df["dataset"].unique():
    print("\n=== ", ds, "===")
    t = df[df["dataset"] == ds][["tracker", "HOTA", "MOTA", "IDF1", "IDSW", "Frag"]].sort_values("HOTA", ascending=False)
    display(t.round(2))
```

### Plot pembanding

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np

dfv = df.copy()
for ds in dfv["dataset"].unique():
    t = dfv[dfv["dataset"] == ds]
    fig, axes = plt.subplots(1, 3, figsize=(14, 4))
    for ax, met in zip(axes, ["HOTA", "IDF1", "MOTA"]):
        ax.bar(t["tracker"], t[met], color=["#4C72B0", "#DD8452"])
        ax.set_title(f"{met} — {ds}")
        for i, v in enumerate(t[met]):
            ax.text(i, v, f"{v:.1f}", ha="center", va="bottom")
    fig.tight_layout()
    out = EXP / f"plot_{ds}.png"
    fig.savefig(out, dpi=150)
    print("saved", out)
```

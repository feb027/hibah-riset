# Fine-Tuning YOLOv10 (Nano) sebagai Baseline Akademik

Notebook ini dirancang untuk melatih (fine-tune) varian **YOLOv10n** menggunakan dataset **CrowdHuman**. 
Model ini akan digunakan sebagai *baseline* pembanding yang adil (*Apples-to-Apples*) melawan YOLO26n di laporan tesis Anda.

```python
# 1. Pastikan library Ultralytics sudah terinstal dan GPU terbaca
import torch
import ultralytics

print("Ultralytics version:", ultralytics.__version__)
print("GPU Available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU Device:", torch.cuda.get_device_name(0))

ultralytics.checks()
```

## Konfigurasi Dataset
Membaca konfigurasi yang sudah kita buat sebelumnya.

```python
import yaml
import os

os.makedirs('/opt/shared_data/hibah-riset/configs', exist_ok=True)

dataset_config = {
    'path': '/opt/shared_data/hibah-riset/data/processed/crowdhuman',
    'train': 'images/train',
    'val': 'images/val',
    'names': {
        0: 'person'
    }
}

with open('/opt/shared_data/hibah-riset/configs/dataset.yaml', 'w') as f:
    yaml.dump(dataset_config, f, default_flow_style=False)

print("File configs/dataset.yaml siap!")
```

## Memulai Training (YOLOv10-Nano)
Kita memuat `yolov10n.pt`. Folder output akan dinamakan `yolo10_crowdhuman` agar tidak menimpa hasil YOLO26.

```python
from ultralytics import YOLO

# Memuat model pre-trained YOLOv10n (Nano)
model = YOLO("yolo11n.pt")  # Ultralytics akan otomatis mendownload jika belum ada

# Memulai proses training
results = model.train(
    data="/opt/shared_data/hibah-riset/configs/dataset.yaml",
    epochs=100,
    imgsz=640,
    batch=32,
    device=0, 
    project="/opt/shared_data/hibah-riset/runs/detect",
    name="yolo11_crowdhuman"  # NAMA FOLDER BARU (TIDAK MENIMPA)
)
```

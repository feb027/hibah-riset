# Fine-Tuning YOLO26 (Small) untuk Real-Time People Counting

Notebook ini dirancang untuk melatih (fine-tune) varian **YOLO26s (Small)** menggunakan dataset **CrowdHuman**. 
Varian Small memiliki parameter lebih banyak dari Nano, diharapkan lebih akurat mengenali objek kecil di kejauhan.

```python
# 1. Pastikan library Ultralytics sudah terinstal dan GPU terbaca
import torch
import ultralytics

print("Ultralytics version:", ultralytics.__version__)
print("GPU Available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU Device:", torch.cuda.get_device_name(0))

ultralytics.checks()
```

## Konfigurasi Dataset
Membaca konfigurasi yang sudah kita buat sebelumnya.

```python
import yaml
import os

os.makedirs('/opt/shared_data/hibah-riset/configs', exist_ok=True)

dataset_config = {
    'path': '/opt/shared_data/hibah-riset/data/processed/crowdhuman',
    'train': 'images/train',
    'val': 'images/val',
    'names': {
        0: 'person'
    }
}

with open('/opt/shared_data/hibah-riset/configs/dataset.yaml', 'w') as f:
    yaml.dump(dataset_config, f, default_flow_style=False)

print("File configs/dataset.yaml siap!")
```

## Memulai Training (YOLO26-Small)
Kita memuat `yolo26s.pt`. Folder output akan dinamakan `yolo_crowdhuman_small` agar tidak menimpa hasil model Nano sebelumnya.

```python
from ultralytics import YOLO

# Memuat model pre-trained YOLO26s (Small)
model = YOLO("/opt/shared_data/hibah-riset/yolo26s.pt")

# Memulai proses training
results = model.train(
    data="/opt/shared_data/hibah-riset/configs/dataset.yaml",
    epochs=100,
    imgsz=640,
    batch=32,
    device=0, 
    project="/opt/shared_data/hibah-riset/runs/detect",
    name="yolo_crowdhuman_small"  # NAMA FOLDER BARU (TIDAK MENIMPA)
)
```
